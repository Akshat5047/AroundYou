## Checking Real Photo Coverage — OpenStreetMap / Wikimedia Commons
Testing how many real Telangana tourist spots have an actual usable
photo reference, before committing to this as the Image Classification
data source.

```python
import requests

TELANGANA_BBOX = "15.8, 77.2, 19.9, 81.3"

query = f"""
[out:json][timeout:60];
(
  node["tourism"]({TELANGANA_BBOX});
  way["tourism"]({TELANGANA_BBOX});
);
out tags;
"""

headers = {
    "User-Agent": "AroundYou-StudentProject/1.0 (contact: your_email@example.com)",
    "Accept": "application/json",
}

response = requests.post(
    "https://overpass-api.de/api/interpreter",
    data={"data": query},
    headers=headers,
    timeout=90,
)

print("Status code:", response.status_code)
print("Response headers:", dict(response.headers))
print("\nActual response body (this should tell us WHY it failed):")
print(response.text[:2000])  # first 2000 characters, in case it's long
```
